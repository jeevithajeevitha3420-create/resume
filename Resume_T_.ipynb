```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load Transformer model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Resume
resume = """
I am an Artificial Intelligence and Machine Learning student.
I know Python, Machine Learning, SQL, TensorFlow and NumPy.
I have worked on AI projects.
"""

# Job description
job = """
We are looking for an AI/ML Engineer with Python,
Machine Learning, SQL and TensorFlow skills.
"""

# Convert resume and job description into embeddings
resume_embedding = model.encode([resume])
job_embedding = model.encode([job])

# Calculate similarity
similarity = cosine_similarity(
    resume_embedding,
    job_embedding
)[0][0]

# Convert to percentage
score = similarity * 100

print(f"Resume-Job Match: {score:.2f}%")
```
